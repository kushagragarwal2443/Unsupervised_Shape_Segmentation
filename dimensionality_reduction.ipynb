```python
from sklearn.cluster import KMeans
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.decomposition import PCA 
from sklearn.preprocessing import StandardScaler
```

```python
def get_reduced_dimension(input_name, dimensions, output_name):

    shapes = np.load(input_name)
    print("Original dimensions:", shapes.shape)
    X_std = shapes
    sklearn_pca = PCA(n_components=dimensions)
    Y = sklearn_pca.fit_transform(X_std)
    print("Reduced dimensions:", Y.shape)
    exp_var_pca = sklearn_pca.explained_variance_ratio_
    cum_sum_eigenvalues = np.cumsum(exp_var_pca)
    print("Total explained variance:", cum_sum_eigenvalues[-1]*100)
    np.save(output_name, Y)
```

```python
get_reduced_dimension("p7_ShapeReps_246x6144.npy", 100, "p7_ShapeReps_246x100")
```
